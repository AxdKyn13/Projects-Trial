```python
import seaborn as sns
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.svm import SVC
from sklearn.metrics import confusion_matrix, accuracy_score

from google.colab import drive
drive.mount('/content/drive')

df = pd.read_csv('/content/drive/MyDrive/trial ML imercy/seattle-weather.csv')

#menghapus kolom, baris, beberapa kolom sekaligus, beberapa baris sekaligus
df = df.drop('date',axis=1)

x = df.drop('weather',axis=1)
y = df['weather']

#Split the dataset into train and test
#25% data dipakai untuk testing, 75% data dipakai untuk training
#Agar pembagian data konsisten setiap kali dijalankan (random state=0)

from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size = 0.25, random_state = 0)
#Feature Scaling = proses menyamakan skala semua fitur (kolom) agar berada dalam rentang yang sebanding, biasanya: Mean = 0 & Standard deviation = 1 (z-score)
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)

#SVM
classifier = SVC(kernel = 'linear', random_state = 0)
classifier.fit(X_train, y_train)

#Predict the tset set result
y_pred = classifier.predict(X_test)

cm = confusion_matrix(y_test, y_pred)
print(cm)
acc2 = accuracy_score(y_test, y_pred)

print(f"Accuracy score: {acc2}")

sns.heatmap(cm,annot=True)
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import confusion_matrix, accuracy_score

from google.colab import drive
drive.mount('/content/drive')

df = pd.read_csv('/content/drive/MyDrive/trial ML imercy/seattle-weather.csv')

#menghapus kolom, baris, beberapa kolom sekaligus, beberapa baris sekaligus
df = df.drop('date',axis=1)

x = df.drop('weather',axis=1)
y = df['weather']

#Split the dataset into train and test
#25% data dipakai untuk testing, 75% data dipakai untuk training
#Agar pembagian data konsisten setiap kali dijalankan (random state=0)

from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size = 0.25, random_state = 0)
#Feature Scaling = proses menyamakan skala semua fitur (kolom) agar berada dalam rentang yang sebanding, biasanya: Mean = 0 & Standard deviation = 1 (z-score)
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)

#Naive Bayes
classifier = GaussianNB()
classifier.fit(X_train, y_train)

#Predict the tset set result
y_pred = classifier.predict(X_test)

cm = confusion_matrix(y_test, y_pred)
print(cm)
acc2 = accuracy_score(y_test, y_pred)

print(f"Accuracy score: {acc2}")

sns.heatmap(cm,annot=True)
plt.show()
```
