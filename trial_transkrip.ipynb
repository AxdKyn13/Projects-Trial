```python
# =========================================
# IMPORT LIBRARY
# =========================================

import whisper
import librosa
import soundfile as sf
import noisereduce as nr

from google.colab import drive
from google.colab import files


# =========================================
# CONNECT GOOGLE DRIVE
# =========================================

drive.mount('/content/drive')


# =========================================
# PATH FILE AUDIO MP3
# =========================================
# GANTI DENGAN LOKASI FILE MP3 KAMU DI DRIVE

audio_path = "/content/drive/MyDrive/NLP trial my/Test2.mp3"


# =========================================
# LOAD AUDIO
# =========================================

audio, sr = librosa.load(audio_path, sr=None)

print("Sample Rate:", sr)
print("Durasi Audio:", len(audio)/sr, "detik")


# =========================================
# NOISE REDUCTION
# =========================================

reduced_noise = nr.reduce_noise(
    y=audio,
    sr=sr
)

clean_audio_path = "/content/drive/MyDrive/NLP trial my/clean_audio_Test2.wav"

sf.write(clean_audio_path, reduced_noise, sr)

print("Noise reduction selesai")


# =========================================
# LOAD WHISPER MODEL
# =========================================
# tiny   = paling cepat
# base   = balance
# small  = lebih akurat
# medium = lebih akurat lagi tapi lebih berat

# Changed from 'large' to 'medium' to reduce memory usage
model = whisper.load_model("medium")


# =========================================
# TRANSCRIBE AUDIO
# =========================================

result = model.transcribe(
    clean_audio_path,
    language = 'id'
)

transcription = result["text"]

# =========================================
# SAVE TO TXT
# =========================================

txt_output = "/content/drive/MyDrive/NLP trial my/hasil_transkrip_Test2.txt"

with open(txt_output, "w", encoding="utf-8") as f:
    f.write(transcription)

print("\nFile txt berhasil dibuat")
```
